```python
#import all the necessary modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#import the dataset and store the data in a variable
data = pd.read_csv("/Users/kanishkasuresh/Documents/FDS LAB/Group assignment/medical_insurance.csv")
df = pd.DataFrame(data)
df
```

```python
#assign arrays to variables x and y, where x contains all the independent values and y contains all the dependent values
X  = df.iloc[:,:-1].values
Y = df.iloc[:,-1].values
df
print(X)
print("\n")
print(Y)
```

```python
#labelencoder transforms non-numericallabels to numerical labels
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

#label encode: sex
X[:,1] = le.fit_transform(X[:,1])

#label encod:e smoker
X[:,4] = le.fit_transform(X[:,4])
```

```python
#one hot encoding: region

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# ColumnTransformer's transformers takes 3 params: operation name,type of operation,column index(s) you want to transform
ct = ColumnTransformer(transformers=[('encoder',OneHotEncoder(),[5])],
                       # passthrough = keep the columns that are not being transformed as it is
                       remainder='passthrough')

# X(feautures) is expected to be a numpy array
X = np.array(ct.fit_transform(X))
```

```python
#split the data into training and testing data after importing the necessary method from sklearn
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
#train the model after importing the necessary method from sklearn
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train,Y_train)
```

```python
#predict the test results
y_pred = regressor.predict(X_test)
y_pred
```

```python
from sklearn.metrics import r2_score
# Calculate R-squared
r2 = r2_score(Y_test, y_pred)
print("R-squared:", r2)
```
